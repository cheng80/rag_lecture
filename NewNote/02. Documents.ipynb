## Document를 직접 만들어 보는 이유

파일 로더를 쓰기 전에 직접 `Document`를 만들어 보면, LlamaIndex가 결국 어떤 형태의 데이터를 기대하는지 알 수 있습니다.
핵심은 본문인 `text`와 부가 정보인 `metadata`를 분리해서 관리하는 것입니다.

```python
from llama_index.core import Document

# Document는 LlamaIndex가 다루는 가장 기본적인 원본 데이터 단위입니다.
# text에는 실제 본문을 넣고, metadata에는 검색/필터링/출처 표시에 필요한 부가 정보를 넣습니다.
text = """
AI 챗봇 사용 가이드:
- 간단한 질문으로 시작하세요.
- 구체적인 예시를 포함하면 더 좋은 답변을 받을 수 있습니다.
- 부적절한 내용은 자동으로 필터링됩니다.
"""

document = Document(
    text=text,
    metadata={
        'author': '제품개발팀',
        'category': '사용설명서',
        'version': '1.0',
    },
    # LlamaIndex의 명시적 문서 ID 필드는 id_입니다.
    # id로 넘기면 버전에 따라 무시되거나 별도 필드로 처리될 수 있어 추적이 불안정합니다.
    id_='manual_ai_001',
)

print(document)
```

```python
# metadata는 답변 근거 표시, 파일명 필터링, 카테고리별 검색 등에 사용됩니다.
document.metadata
```

## 좋은 metadata의 기준

metadata에는 검색 결과를 설명하거나 필터링할 때 필요한 값만 넣는 것이 좋습니다.
예를 들어 `file_name`, `category`, `version`, `page_label` 같은 값은 유용하지만, 너무 긴 설명문이나 중복 본문을 넣으면 검색 노이즈가 늘 수 있습니다.

### 2. 기준 문서에서 Document 객체 생성

```python
from llama_index.core import SimpleDirectoryReader

# 실제 파일에서 Document를 생성하는 예제입니다.
# PDF 하나가 반드시 Document 하나가 되는 것은 아니며, 리더 설정에 따라 페이지/파일 단위가 달라질 수 있습니다.
documents = SimpleDirectoryReader(input_dir='../NewData/pdf_sample1').load_data()
print('읽어온 문서 수:', len(documents))
```

## 직접 생성한 Document와 파일에서 읽은 Document 비교

직접 만든 Document는 구조를 이해하기 좋고, 파일에서 읽은 Document는 실제 RAG 파이프라인에 가깝습니다.
두 방식 모두 결과가 `Document` 목록이라는 점이 중요합니다.
즉, 로더가 달라도 이후 인덱싱 코드는 거의 같은 방식으로 이어집니다.

```python
# 첫 번째 Document 본문 일부 확인
print(documents[0].text[:1000])
```

```python
# 파일에서 자동으로 생성된 메타데이터 확인
# - file_name은 출처 표시나 특정 파일 필터링에 자주 사용합니다.
print(documents[0].metadata)
print('file_name:', documents[0].metadata.get('file_name'))
```
